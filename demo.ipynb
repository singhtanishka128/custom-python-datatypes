### imports

```python
# Automatically reload imported modules if you change the .py files
%load_ext autoreload
%autoreload 2

# Import your custom datatypes
from complex_number import ComplexNumber
from distance import Distance
from interval import Interval
from matrix import Matrix
from polynomial import Polynomial

print("All custom datatypes imported successfully!")
```

### test complex numbers

```python
c1 = ComplexNumber(3, 2)
c2 = ComplexNumber(1, 7)

print(f"c1: {c1}")
print(f"c2: {c2}")
print(f"Addition (c1 + c2): {c1 + c2}")
print(f"Multiplication (c1 * c2): {c1 * c2}")
print(f"Magnitude of c1: {abs(c1):.2f}")
```

### test matrices

```python
m1 = Matrix([[1, 2], [3, 4]])
m2 = Matrix([[5, 6], [7, 8]])

print("Matrix 1:")
print(m1)
print("\nMatrix Multiplication (m1 @ m2):")
print(m1 @ m2)
print("\nTranspose of m1:")
print(m1.transpose())
```

### test intervals

```python
i1 = Interval(1, 5)
i2 = Interval(3, 8)

print(f"Interval 1: {i1}")
print(f"Is 3 in {i1}? {3 in i1}")
print(f"Do {i1} and {i2} overlap? {i1.overlaps(i2)}")
print(f"Intersection: {i1.intersection(i2)}")
```

### test polynomials

```python
# 3x^2 + 2x + 5
poly = Polynomial([3, 2, 5])
print(f"Polynomial: {poly}")
print(f"Evaluated at x = 2: {poly(2)}")

poly2 = Polynomial([1, 4])  # x + 4
print(f"Poly 1 + Poly 2: {poly + poly2}")
```

### test distances

```python
d1 = Distance(5, "m")
d2 = Distance(10, "ft")

print(f"Distance 1: {d1}")
print(f"Distance 2 converted to meters: {d2.to('m')}")
print(f"Sum (d1 + d2): {d1 + d2}")
```
